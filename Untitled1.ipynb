```python
food = {
    "Product": ["Pizza", "Burger", "Salad", "Chicken", "Pasta",
                "Fish", "Rice", "Sandwich", "Soup", "Steak"],
    "Category": ["Fast Food", "Fast Food", "Healthy", "Protein",
                 "Fast Food", "Protein", "Grain", "Fast Food",
                 "Healthy", "Protein"],
    "Calories": [450, 550, 180, 320, 420, 280, 350, 400, 150, 500],
    "Protein": [15, 25, 5, 35, 12, 30, 7, 18, 6, 40]
}

df_food = pd.DataFrame(food)

df_food
```

```python
df_food[df_food["Calories"] > 300]
```

```python
df_food.groupby("Category")["Protein"].mean()
```

```python
df_food.sort_values(by="Calories", ascending=False)
```

```python
employees = {
    "Name": ["Ali", "Aysel", "Murad", "Nigar",
             "Ali", "Aysel", "Murad", "Nigar"],
    "Project": ["Website", "Mobile App", "Website", "Database",
                "Mobile App", "Website", "Database", "Mobile App"],
    "Hours": [15, 20, 12, 18, 10, 25, 14, 16]
}

df_employees = pd.DataFrame(employees)

df_employees
```

```python
df_employees.groupby("Name")["Hours"].sum()
```

```python
df_employees.groupby("Project")["Hours"].sum()
```

```python
df_employees.groupby("Name")["Hours"].sum().sort_values(ascending=False).head(1)
```

```python
tickets = {
    "Movie": ["Avatar", "Titanic", "Avatar", "Joker",
              "Titanic", "Joker", "Avatar", "Titanic",
              "Joker", "Avatar", "Titanic", "Joker"],
    "City": ["Baku", "Ganja", "Baku", "Sumqayit",
             "Baku", "Ganja", "Ganja", "Sumqayit",
             "Baku", "Sumqayit", "Ganja", "Baku"],
    "TicketsSold": [120, 90, 150, 80, 110, 100,
                    130, 70, 140, 95, 85, 115]
}

df_tickets = pd.DataFrame(tickets)

df_tickets
```

```python
df_tickets.groupby("Movie")["TicketsSold"].sum()
```

```python
df_tickets.groupby("City")["TicketsSold"].sum()
```

```python
movie_sales = df_tickets.groupby("Movie")["TicketsSold"].sum()

movie_sales
```

```python
movie_sales.sort_values(ascending=False).head(1)
```
